```python
# 202431191_M. Sidiq Maulana
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 202431191_M. Sidiq Maulana
data = pd.read_csv("drug200.csv")
data.head()
```

```python
# 202431191_M. Sidiq Maulana
data.info()
```

```python
# 202431191_M. Sidiq Maulana
data.describe()
```

```python
# 202431191_M. Sidiq Maulana
numerical_features = features = [feature for feature in data.columns if data[feature].dtype != 'O']
print("Jumlah fitur numerik = ",
      len(numerical_features))
print("Daftar fitur numerik = ",
      numerical_features)
display(data[numerical_features].head())
```

```python
# 202431191_M. Sidiq Maulana
categorical_features = features = [feature for feature in data.columns if data[feature].dtype == 'O']
print("Jumlah fitur kategorikal = ",
      len(categorical_features))
print("Daftar fitur kategorikal = ",
      categorical_features)
display(data[categorical_features].head())
```

```python
# 202431191_M. Sidiq Maulana
for i in categorical_features:
  print(data[i].unique())
```

```python
# 202431191_M. Sidiq Maulana
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for feature in categorical_features:
  data[feature] = le.fit_transform(data[feature])
print(data[feature].unique)
```

```python
# 202431191_M. Sidiq Maulana
x = data.drop(['Drug'], axis = 1)
y = data['Drug']
print(x)
print(y)
```

```python
# 202431191_M. Sidiq Maulana
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

```python
# 202431191_M. Sidiq Maulana
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(x_train, y_train)
```

```python
# 202431191_M. Sidiq Maulana
y_pred_train = model.predict(x_train)
y_pred_test = model.predict(x_test)
```

```python
# 202431191_M. Sidiq Maulana
from sklearn.metrics import confusion_matrix, classification_report

print("Evaulasi data training")
cm_train = confusion_matrix(y_train, y_pred_train)
print(cm_train)
print(classification_report(y_train, y_pred_train))

plt.figure(figsize=(6,4))
sns.heatmap(cm_train, annot=True, fmt='d', cmap='Blues',
            xticklabels=np.unique(y), yticklabels=np.unique(y))
plt.title('Confusion Matrix - Training')
plt.xlabel('Label Terprediksi')
plt.ylabel('Label Aktual')
plt.show()
```
